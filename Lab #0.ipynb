#### Electric Force Between Two charges
$$ F_e =  \frac{q_1 * q_2}{4 \pi \epsilon_0 r^2}$$

```python
import math
```

```python
eps = 8.9e-12

def Fe(q1,q2,r):
  F = q1 * q2 / (4 * math.pi * eps * r**2)
  return F

print(f"{Fe(1,1,1):.2e}")
```

#### escape velocity
$$v_{esc} = \sqrt{\frac{2GM}{R}}$$

```python
G = 6.67e-11
M = 1.98e30
R = 6.37e6

def vesc(m,r):
  v = math.sqrt(2 * G * m / r)
  return v

print(f"{vesc(M,R):.2e}")
```

```python
from scipy.integrate import solve_ivp

# Define ODE
def dydt(t, y):
    k = 0.5
    return -k * y

# Solve ODE
solution = solve_ivp(dydt, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print("Final value:", solution.y[0][-1])
```

### Gradient Descent

![link text](https://baptiste-monpezat.github.io/6f5cc892ec345d96f64c881b62b0d910/gradient_descent_parameter_a.gif)

> In physics, systems evolve toward minimum energy.
> In machine learning, models evolve toward minimum error.


We want to minimize a function.

Example: Fit a straight line

$$
y = mx + b
$$

Define loss (Mean Squared Error):

$$
L(m,b) = \frac{1}{N} \sum_{i=1}^N (y_i - y^{pred}_i)^2
$$

For:

$$
y^{pred} = mx_i + b
$$

$$
L(m,b) = \frac{1}{N} \sum_{i=1}^N (y_i - (mx_i + b))^2
$$

We want:

$$
\min_{m,b} L(m,b)
$$


Gradient = direction of steepest increase.

$$
m := m - \alpha \cdot \frac{\partial L}{\partial m}
$$

$$
b := b - \alpha \cdot \frac{\partial L}{\partial b}
$$

The gradients are:

$$
\frac{\partial L}{\partial m} = -\frac{2}{n} \sum_{i=1}^{n} x_i (y_i - (m x_i + b))
$$

$$
\frac{\partial L}{\partial b} = -\frac{2}{n} \sum_{i=1}^{n} (y_i - (m x_i + b))
$$


If learning rate is:

* Too small → slow convergence
* Too large → oscillation or divergence

![link text](https://miro.medium.com/v2/resize:fit:1100/format:webp/1*Nq6UMjR0WyH0Q8sDYmSzaQ.gif)


```python
# Variables
x = np.linspace(0, 10, 10)
y = 3*x + 2 + np.random.normal(0, 2, 10)

# Initialize parameters
a = 0
b = 0
lr = 0.01
epochs = 1000
```

```python
# Mean Squared Error loss.
def compute_loss(m, b, X, y):
    n = len(y)
    predictions = m * X + b
    loss = np.sum((y - predictions) ** 2) / n
    return loss
```

```python
def gradient_descent(X, y, m, b, learning_rate, epochs):
    """
    Perform gradient descent to update m and b.
    Returns updated m, b, and history of losses.
    """
    n = len(y)
    loss_history = []

    for i in range(epochs):
        # Predictions
        y_pred = m * X + b

        # Compute gradients
        dm = (-2 / n) * np.sum(X * (y - y_pred))
        db = (-2 / n) * np.sum(y - y_pred)

        # Update parameters
        m = m - learning_rate * dm
        b = b - learning_rate * db

        # Record loss
        current_loss = compute_loss(m, b, X, y)
        loss_history.append(current_loss)

    return m, b, loss_history
```

```python
# Train the model
m_final, b_final, loss_hist = gradient_descent(x, y, a, b, lr, epochs)

print(f"Final Parameters:")
print(f"Slope (m): {m_final}")
print(f"Intercept (b): {b_final}")
```

```python
# Plot Training Loss Over Time

loss_hist = [float(l) for l in loss_hist]

plt.figure(figsize=(8, 5))
plt.plot(loss_hist, color='red')
plt.title('Training Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Mean Squared Error (MSE)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Plot
plt.figure(figsize=(8,6))

# Scatter data
plt.scatter(x, y, label="Data Points")

# Final fitted line
plt.plot(x, m_final*x + b_final, label="Fitted Line (Gradient Descent)")

# True line
plt.plot(x, 3*x + 2, label="True Line")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression using Gradient Descent")

plt.legend()
plt.grid(True)

plt.show()
```
